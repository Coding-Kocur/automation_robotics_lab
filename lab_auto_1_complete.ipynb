```python
# Basics of Pulp

import pulp

# Creating a problem
prob = pulp.LpProblem("Simple_Problem", pulp.LpMaximize) # we want to maximize the objective function

# Variables
x = pulp.LpVariable('x', lowBound=0, upBound=4) # x must be between 0 and 4
y = pulp.LpVariable('y', lowBound=0, upBound=3) # y must be between 0 and 3

# Objective function
prob += x + y, "Z" # we want to maximize the sum of x and y

# Constraints
prob += x - y == 0 # they must be same
print(prob)
```

```python
# Solving the problem
prob.solve()

# Printing the results
print("x =", x.value())
print("y =", y.value())
```

```python
# Whiskas Blending Problem with sardines and vitamin B constraint
import pulp

# Ingredients
Ingredients = ['CHICKEN', 'BEEF', 'MUTTON', 'RICE', 'WHEAT', 'GEL', 'SARDINES']

# Costs per gram
costs = {'CHICKEN': 0.013, 'BEEF': 0.008, 'MUTTON': 0.010,
         'RICE': 0.002, 'WHEAT': 0.005, 'GEL': 0.001, 'SARDINES': 0.009}

# Protein percent
proteinPercent = {'CHICKEN': 0.100, 'BEEF': 0.200, 'MUTTON': 0.150,
                  'RICE': 0.000, 'WHEAT': 0.040, 'GEL': 0.000, 'SARDINES': 0.180}

# Fat percent
fatPercent = {'CHICKEN': 0.080, 'BEEF': 0.100, 'MUTTON': 0.110,
              'RICE': 0.010, 'WHEAT': 0.010, 'GEL': 0.000, 'SARDINES': 0.090}

# Fibre percent
fibrePercent = {'CHICKEN': 0.001, 'BEEF': 0.005, 'MUTTON': 0.003,
                'RICE': 0.100, 'WHEAT': 0.150, 'GEL': 0.000, 'SARDINES': 0.002}

# Salt percent
saltPercent = {'CHICKEN': 0.002, 'BEEF': 0.005, 'MUTTON': 0.007,
               'RICE': 0.002, 'WHEAT': 0.008, 'GEL': 0.000, 'SARDINES': 0.003}

# Vitamin B content in μg/100g
vitaminB = {'CHICKEN': 0.3, 'BEEF': 2.6, 'MUTTON': 2.6,
            'RICE': 0.0, 'WHEAT': 0.0, 'GEL': 0.0, 'SARDINES': 8.9}

# Create the problem
prob = pulp.LpProblem("Whiskas_Model", pulp.LpMinimize)

# Decision variables (percentage of each ingredient)
ingredient_vars = pulp.LpVariable.dicts("Ingr", Ingredients, 0)

# Objective: minimize cost
prob += pulp.lpSum([costs[i] * ingredient_vars[i] for i in Ingredients]), "Total Cost per can"

# Constraints
prob += pulp.lpSum([ingredient_vars[i] for i in Ingredients]) == 100, "PercentagesSum"
prob += pulp.lpSum([proteinPercent[i] * ingredient_vars[i] for i in Ingredients]) >= 8.0, "ProteinRequirement"
prob += pulp.lpSum([fatPercent[i] * ingredient_vars[i] for i in Ingredients]) >= 6.0, "FatRequirement"
prob += pulp.lpSum([fibrePercent[i] * ingredient_vars[i] for i in Ingredients]) <= 2.0, "FibreRequirement"
prob += pulp.lpSum([saltPercent[i] * ingredient_vars[i] for i in Ingredients]) <= 0.4, "SaltRequirement"

# Vitamin B constraint: minimum 0.8 μg per 100g of food
prob += pulp.lpSum([vitaminB[i] * ingredient_vars[i] for i in Ingredients]) >= 80.0, "VitaminBRequirement"

# Solve
prob.solve(pulp.PULP_CBC_CMD(msg=0))

print("Status:", pulp.LpStatus[prob.status])
print("\nIngredient percentages:")
for v in prob.variables():
    if v.varValue > 0:
        print(f"  {v.name} = {v.varValue:.2f}%")
print(f"\nTotal cost per 100g = {pulp.value(prob.objective):.4f}")
```
